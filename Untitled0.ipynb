# Cancer Variant Prioritization Machine Learning Project

In this notebook, we utilize the `/cancer_variant_prioritization_dataset (1).csv` dataset to build a predictive model. We will inspect the data structure, handle preprocessing, train a classifier, and evaluate the performance.

### 5. Preprocessing and Feature Encoding
To feed our categorical features into a machine learning model, we will use label encoders for our string-based features (`Gene`, `Variant_Type`, `SIFT`, `PolyPhen`) and target classes.

```python
# Make a copy of our data
processed_df = df_variant.copy()

# Identify categorical columns to encode
categorical_cols = ['Gene', 'Variant_Type', 'SIFT', 'PolyPhen']

# Instantiate label encoders
encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    processed_df[col] = le.fit_transform(processed_df[col])
    encoders[col] = le

# Encode target variable
target_encoder = LabelEncoder()
processed_df['Clinical_Significance'] = target_encoder.fit_transform(processed_df['Clinical_Significance'])

print("Encoded Classes for Target:")
for idx, class_name in enumerate(target_encoder.classes_):
    print(f"  {idx} -> {class_name}")

# Preview the preprocessed dataset
display(processed_df.head())
```

### 6. Training the Predictive Model
We will now partition the dataset into training (80%) and testing (20%) sets, train a Random Forest Classifier, and measure its prediction capability.

```python
# Define features and target
features = ['Chromosome', 'Position', 'Gene', 'Variant_Type', 'Population_AF', 'CADD_Score', 'SIFT', 'PolyPhen']
X_var = processed_df[features]
y_var = processed_df['Clinical_Significance']

# Perform split
X_train_v, X_test_v, y_train_v, y_test_v = train_test_split(
    X_var, y_var, test_size=0.2, random_state=42, stratify=y_var
)

# Train Random Forest Classifier
rf_model_v = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model_v.fit(X_train_v, y_train_v)

# Evaluate predictions
y_pred_v = rf_model_v.predict(X_test_v)
accuracy_v = accuracy_score(y_test_v, y_pred_v)

print(f"Model Accuracy on Variant Prioritization: {accuracy_v:.4f}\n")
print("Classification Report:")
print(classification_report(y_test_v, y_pred_v, target_names=target_encoder.classes_))

# Plot confusion matrix
cm_v = confusion_matrix(y_test_v, y_pred_v)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_v, annot=True, fmt='d', cmap='Purples', xticklabels=target_encoder.classes_, yticklabels=target_encoder.classes_)
plt.title('Confusion Matrix - Variant Prioritization Model')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

### 8. Advanced Feature Engineering & Data Enrichment

Since raw label encoding yielded random performance, we will now transform our columns based on domain knowledge:
- **SIFT & PolyPhen Mapping:** Convert qualitative predictions to numerical impact ranks.
- **Gene Frequency Encoding:** Encode genes by their frequency of occurrence in the dataset to represent mutational hotspots.
- **Interaction Features:** Combine the computational pathogenicity metric (`CADD_Score`) with the rarity of the mutation (`Population_AF`).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
import xgboost as xgb

# Start with a fresh copy of the original dataset
fe_df = df_variant.copy()

# 1. Map SIFT and PolyPhen to ordinal scores based on severity
# Let's check unique values first to ensure we map them correctly
print("Unique SIFT:", fe_df['SIFT'].unique())
print("Unique PolyPhen:", fe_df['PolyPhen'].unique())

sift_map = {'deleterious': 2, 'deleterious_low_confidence': 1, 'tolerated_low_confidence': 0.5, 'tolerated': 0}
polyphen_map = {'probably_damaging': 2, 'possibly_damaging': 1, 'benign': 0}

fe_df['SIFT_numeric'] = fe_df['SIFT'].map(sift_map).fillna(0)
fe_df['PolyPhen_numeric'] = fe_df['PolyPhen'].map(polyphen_map).fillna(0)

# 2. Gene Frequency Encoding
gene_freq = fe_df['Gene'].value_counts() / len(fe_df)
fe_df['Gene_Freq'] = fe_df['Gene'].map(gene_freq)

# 3. Create Interaction Features
# Variants with low allele frequency (rare) and high CADD scores are more likely to be pathogenic
fe_df['Pathogenicity_Index'] = fe_df['CADD_Score'] * (1 - fe_df['Population_AF'])

# 4. One-Hot Encode Variant_Type
fe_df = pd.get_dummies(fe_df, columns=['Variant_Type'], drop_first=True)

# 5. Drop unneeded raw columns
cols_to_drop = ['Gene', 'SIFT', 'PolyPhen']
fe_df = fe_df.drop(columns=cols_to_drop)

# Display enriched dataframe features
display(fe_df.head())
```

### 9. Retraining XGBoost on Enriched Features

```python
# Encode the target
target_encoder_fe = LabelEncoder()
fe_df['Clinical_Significance'] = target_encoder_fe.fit_transform(fe_df['Clinical_Significance'])

# Define new engineered feature space
fe_features = [col for col in fe_df.columns if col != 'Clinical_Significance']

X_fe = fe_df[fe_features]
y_fe = fe_df['Clinical_Significance']

# Split
X_train_fe, X_test_fe, y_train_fe, y_test_fe = train_test_split(
    X_fe, y_fe, test_size=0.2, random_state=42, stratify=y_fe
)

# Train improved XGBoost model
optimized_xgb = xgb.XGBClassifier(
    n_estimators=150,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='mlogloss'
)

optimized_xgb.fit(X_train_fe, y_train_fe)

# Evaluate
y_pred_fe = optimized_xgb.predict(X_test_fe)
accuracy_fe = accuracy_score(y_test_fe, y_pred_fe)

print(f"Optimized XGBoost Accuracy: {accuracy_fe:.4f}\n")
print(classification_report(y_test_fe, y_pred_fe, target_names=target_encoder_fe.classes_))
```

### 10. Diagnostic Analysis: Assessing Feature Signals

Because both Random Forest and XGBoost models perform around ~25% accuracy (exactly matching random selection among 4 classes), we will run a statistical diagnostic using **Mutual Information** and compare performance against a **Dummy Classifier** baseline.

### 11. Detailed Interpretation of Diagnostic Metrics

By checking these critical information-theoretic and baseline measurements, we can conclusively establish the following:

1. **Random Performance Against Baseline:** Our trained Random Forest (~24.95%) and XGBoost (~25.40%) models operate within standard statistical variation of the **Stratified Dummy Classifier** (~24.25%). In a perfectly balanced 4-class problem, random coin flips provide exactly a 25.0% accuracy rate.
2. **Absence of Feature-Target Dependency:** The **Mutual Information (MI)** score measures how much information a feature provides about the target variable. An MI score of exactly `0.0` represents complete statistical independence. In our dataset, features like `CADD_Score`, `Position`, and `Population_AF` scored exactly `0.000000`, with all remaining features trending extremely close to 0.
3. **Uniform Feature Importance:** The relative importances of our variables are distributed almost uniformly (each at roughly 7% to 9%). This indicates that our model is treating all inputs as equivalent random noise and is struggling to assign structured predictive significance to any subset.

**Conclusion:** The clinical classifications in this dataset are mathematically independent of the variant features, signifying that the labels are synthetic or randomized. However, our preprocessing, feature engineering, and training pipeline have been successfully built and are ready to deploy on non-random data.

```python
from sklearn.dummy import DummyClassifier
from sklearn.feature_selection import mutual_info_classif

print("=== 1. Dummy Classifier Baseline ===")
dummy_clf = DummyClassifier(strategy='stratified', random_state=42)
dummy_clf.fit(X_train_fe, y_train_fe)
dummy_pred = dummy_clf.predict(X_test_fe)
dummy_acc = accuracy_score(y_test_fe, dummy_pred)
print(f"Stratified Dummy Classifier Accuracy: {dummy_acc:.4f}")

print("\n=== 2. Mutual Information Scores ===")
mi_scores = mutual_info_classif(X_fe, y_fe, random_state=42)
mi_series = pd.Series(mi_scores, index=X_fe.columns).sort_values(ascending=False)
print("Mutual Information between features and Clinical Significance:")
print(mi_series)

print("\n=== 3. Model Feature Importances (Optimized XGBoost) ===")
xgb_importances = pd.Series(optimized_xgb.feature_importances_, index=X_fe.columns).sort_values(ascending=False)
print(xgb_importances)
```
